**Data description with Pandas** is about understanding the variables or features in the DataFrame.<br>
Pandas functions for creating Descriptive Statistics or making Data Descriptions simply create parametric descriptions of the variables, and we can print them as collections or tables of numbers which are useful for describing the data.

```python
import pandas as pd

cardata = {"Mercedes": [2, 4, 0, 4, 0, 3],
           "Ford": [3, 0, 0, 1, 6, 12],
           "Tata": [9, 3, 4, 1, 0, 0],
           "Renault": [12, 1, 0, 0, 3, 1],
          }
carsales = pd.DataFrame(cardata)
carsales.index.rename("Sales place", inplace=True)
carsales.rename(index={0: "One",
                       1: "Two",
                       2: "Three",
                       3: "Four",
                       4: "Five",
                       5: "Six",
                      },
                inplace=True)
carsales
```

# `.agg()`

- make some summary data descriptors

```python
carsales.agg(["sum", "max", "min"])
```

```python
carsales.agg(sum=("Ford", "sum"), max=("Ford", "max"), min=("Ford", "min"))
```

- make different summary data descriptions for selected columns

```python
carsales.agg({"Ford": ["sum", "max", "min"],
              "Tata": ["sum", "max", "min", "count"],
             })
```

- create aggregate statistics for columns (total number of sales per brand)

```python
carsales[["Mercedes", "Ford", "Tata", "Renault"]].agg("sum")
```

- create aggregate statistics for rows (mean number of sales per salesplace)

```python
carsales.agg("mean", axis=1)
```

# `.describe()`

This function produces a standardized output with parameters in the float format for the DataFrame.<br>
These are:
1. *count*
2. *mean*
3. *standard deviation*
4. *min*
5. *quartiles (25, median, 75)*
6. *max*

```python
carsales.describe()
```

# `.value_counts()`

- show the count of each data item. This can augment descriptive statistics.

```python
carsales["Ford"].value_counts()
```

- **normalize** the value counts to show the % of the distribution values

```python
carsales["Ford"].value_counts(normalize=True)
```

- create a new column with total sales and add it to the DataFrame

```python
carsales.columns
```

```python
carsales["TotalSales"] = carsales[carsales.columns].sum(axis=1)
carsales
```

# `.apply()`

Useful to apply a Python (or any other language) function on every column entry in a DataFrame.

```python
def rate_sales(sold):
    if sold >= 15:
        return "Excellent"
    elif sold >= 5:
        return "Sufficient"
    else:
        return "Insufficient"
```

```python
carsales["SalesRating"] = carsales["TotalSales"].apply(rate_sales)
carsales
```
