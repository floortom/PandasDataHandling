```python
import pandas as pd

cardata = {"Mercedes": [2, 4, 0, 4, 0, 3],
           "Ford": [3, 0, 0, 1, 6, 12],
           "Tata": [9, 3, 4, 1, 0, 0],
           "Renault": [12, 1, 0, 0, 3, 1],
          }
carsales = pd.DataFrame(cardata)
carsales.index.rename("Sales place", inplace=True)
carsales.rename(index={0: "One",
                       1: "Two",
                       2: "Three",
                       3: "Four",
                       4: "Five",
                       5: "Six",
                      },
                inplace=True)
carsales
```

# `.sort_values()`

**Note!** we do not use `inplace=True`

- sort on one column

```python
carsales.sort_values(by=["Ford"])
```

```python
carsales.sort_values(by=["Ford"], ascending=False)
```

- Sort on several columns and also hierarchical structures by increasing the number of `by=` columns.<br>
For this non-hierarchical dataset, such a sort has little meaning.

```python
carsales.sort_values(by=["Ford", "Tata"], ascending=False)
```

# `.sort_index()`

- data gets sorted alphabetically

```python
carsales.sort_index()
```

# `.idxmax()`

- find the rows with the largest values for each column

```python
carsales.idxmax()
```

- find the columns with the largest value for each row

```python
carsales.idxmax(axis="columns")
```

# `.idxmin()`

- find the rows with the smallest values for each column

```python
carsales.idxmin()
```

- find the columns with the smallest values for each row

```python
carsales.idxmin(axis="columns")
```

# Dealing with duplicated max and min values

- Find the 3 rows with largest Mercedes sales, keep all duplicates

```python
carsales.nlargest(3, "Mercedes", keep="all")
```

- Find the 3 lowest Renault sales, keep all duplicates

```python
carsales.nsmallest(3, "Mercedes", keep="all")
```

# `.rank()`

- rank the salesplaces per single column (which salesplace sells the most of the Mercedes brand)

```python
carsales["Mercedes"].rank(ascending=False)
```

- rank the salesplaces per column (which salesplace sales the most of a brand)

```python
carsales.rank(ascending=False)
```

- rank the columns per salesplace (which brand sells the most cars for a salesplace) 

```python
carsales.rank(axis=1, ascending=False)
```
