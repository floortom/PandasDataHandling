# Create datasets with duplicates

```python
import pandas as pd

cardata = {"Mercedes": [2, 4, 0, 4, 0, 3],
           "Ford": [3, 0 , 13, 1, 6, 12],
           "Tata": [9, 3, 4, 1, 0, 0],
           "Renault": [12, 1, 0, 0, 3, 1],
          }
carsales = pd.DataFrame(cardata)
carsales.index.rename("Sales place", inplace=True)
carsales.rename(index={0: "One",
                       1: "Two",
                       2: "Three",
                       3: "Four",
                       4: "Five",
                       5: "Six",
                      },
                inplace = True)
```

```python
places = ["Europe 1", "Australia 1", "USA 1", "Asia 1", "Africa 1", "South America 1"]
carsales.insert(0, "SalesPlaceName", places, allow_duplicates=True)

carsales
```

```python
nsp1 = {"SalesPlaceName": "South America 1",
        "Mercedes": 3,
        "Ford": 2,
        "Tata":1,
        "Renault": 1,
       }
newSalesPlace1 = pd.DataFrame(nsp1, index=["Seven"])
newSalesPlace1
```

```python
nsp2 = {"SalesPlaceName": "Asia 1",
        "Mercedes": 4,
        "Ford": 1,
        "Tata": 1,
        "Renault": 0,
       }
newSalesPlace2 = pd.DataFrame(nsp2, index=["Eight"])
newSalesPlace2
```

## Final dataset with duplicates

There are two different duplicates:
1. One **key-based** duplicate - *South America 1* (only the sales place name)
2. One complete **data** duplicate - *Asia 1* (all values)

```python
carsales = pd.concat([carsales, newSalesPlace1, newSalesPlace2])
carsales
```

# Pandas' `.duplicated()` function

can find **key-based** duplicates in a DataFrame by limiting the duplicates to the key column(s)<br>
`keep=first/last` may determine which duplicates to keep in the DataFrame or an attached DataFrame<br>
`keep=False` marks all duplicates.

```python
carsales.duplicated(subset=["SalesPlaceName"])
```

```python
carsales.duplicated(subset=["SalesPlaceName"], keep="last")
```

```python
carsales.duplicated(subset=["SalesPlaceName"], keep=False)
```

## Complete duplicates

```python
carsales.duplicated()
```

# Pandas' `.drop_duplicates()` function

```python
carsalesDupRem = carsales.drop_duplicates(subset=["SalesPlaceName"], keep="first")
carsalesDupRem
```

## More `subset` columns

useful in a multi-index environment

```python
carsalesDupRem2 = carsales.drop_duplicates(subset=["SalesPlaceName", "Mercedes"], keep="last")
carsalesDupRem2
```

```python
carsalesDupRem3 = carsales.drop_duplicates(subset=["SalesPlaceName", "Ford"], keep="last")
carsalesDupRem3
```
