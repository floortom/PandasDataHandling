```python
import pandas as pd

cardata = {"Mercedes": [2, 4, 0, 4, 0, 3],
           "Ford": [3, 0, 0, 1, 6, 12],
           "Tata": [9, 3, 4, 1, 0, 0],
           "Renault": [12, 1, 0, 0, 3, 1],
          }
carsales = pd.DataFrame(cardata)
carsales.index.rename("Sales place", inplace=True)
carsales.rename(index={0: "One",
                       1: "Two",
                       2: "Three",
                       3: "Four",
                       4: "Fiverr",
                       5: "Six",
                      },
                inplace=True)
carsales
```

# Print column names and index labels

```python
carsales.columns
```

```python
carsales.index
```

# Rename columns

## Using a `.rename()` with a dictionary

```python
carsales.rename(columns={"Tata": "Tatra"}, inplace=True)
carsales
```

## Using `.columns.str.replace()`

```python
carsales.columns = carsales.columns.str.replace("Tatra", "Tatrra")
carsales
```

## Replace the entire set of column strings

```python
carsales.columns = ["Mercedes", "Ford", "Tata", "Renault"]
carsales
```

# Edit index labels

## Using a `.rename()` with a dictionary

```python
carsales.rename(index={"Fiverr": "Five"}, inplace=True)
carsales
```

## Using  `.index.str.replace()`

```python
carsales.index = carsales.index.str.replace("Five", "Fiverr")
carsales
```

## Replace the entire index set 

This also removes the *index name* set up after DF creation

```python
carsales.index = [0, 1, 2, 3, 4, 5]
carsales
```

```python
carsales.index.rename("Sales place", inplace=True)
carsales
```
