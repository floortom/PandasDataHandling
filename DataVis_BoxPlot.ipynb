```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the DataFrame from the *Seaborn* library

```python
iris = sns.load_dataset("iris")
iris
```

# Matplotlib `.boxplot()`

## Matplotblib **box plot** for a single column

- `showmeans` shows the arithmetic mean.

```python
ax = iris.boxplot(column="sepal_length", showmeans=True, meanline=True, grid=False)
plt.suptitle("") # Removes auto-generated plot title
ax.set_title("Box plot showing sepal length for 150 flowers", fontsize=16)
plt.show()
```

## Single-column **box plot** split by the *species* feature

- one for every flower type

```python
ax = iris.boxplot(column="sepal_length", by="species", showmeans=True, meanline=True, grid=False)
plt.suptitle("")
ax.set_title("Box plot showing sepal lengths for 3 flower species", fontsize=16)
ax.set_xlabel("Species", fontsize=12)
plt.show()
```

# Seaborn box plot

- `dodge` parameter affects how the underlying algorithm positions the boxes - at times these may be visually out of line  

```python
sns.boxplot(data=iris, x="species", y="sepal_length", hue="species", dodge=False, width=0.3)
plt.show()
```

## Seaborn `boxenplot`

- Split on the *species* feature.<br>
The Boxen plot creates **more** boxes than the ordinary Box Plot and **adapts them to the local distribution of the variable**. Boxen plot magnifies extreme datapoints.

```python
sns.boxenplot(data=iris, x="species", y="sepal_length", hue="species", dodge=False, width=0.4)
plt.show()
```

## Seaborn `violinplot`

- Split on the *species* feature.<br>
The **Violin Plot** creates a more **density-like** or **distribution-like** plot and increases the information about the variable distribution.

```python
sns.violinplot(data=iris, x="species", y="sepal_length", hue="species", dodge=False, width=0.5)
plt.show()
```

## Seaborn `swarmplot`

These plots create an artificial cloud or swarm of data points along the y-axis to show the location and distribution of the data points.<br>
This creates a better view of the distribution of values, but may not scale well with datasets containing large numbers of data points. 

```python
sns.swarmplot(data=iris, x="species", y="sepal_length", hue="species")
plt.show()
```

***

- The **swarm plot** can be a good complement to a boxen or violin plot.<br>
In cases where all observations need to be shown along with some representation of the underlying data distribution.

```python
sns.catplot(data=iris, x="species", y="sepal_length", hue="species", kind="boxen", dodge=False, width=0.5)
sns.swarmplot(data=iris, x="species", y="sepal_length", hue="species")
plt.show()
```
