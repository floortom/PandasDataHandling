**Indicator features** or so-called **dummy variables** have many uses.<br> 
They are effectively about transforming a *binned* or *categorical* **feature** or **variable** into a new set of features that indicate if a **state or value** of the original feature is **True** or **False**, one or zero for every row or data point.<br>
The main use of indicator features or dummy variables is in modelling and an indicator feature is considered to make you able to **quantify the impact** which an indicated state or feature has on a modelled phenomenon.

# Pandas' `get_dummies()`

## Create a DF with a *region* feature

```python
medCost2 = pd.get_dummies(medCost, columns=["region"], dtype=bool)
medCost2
```

- For modeling purposes, it is more useful to use **integers instead of booleans** for indicator features.

```python
medCost3 = pd.get_dummies(medCost, columns=["region"], dtype=int)
medCost3
```

**ones** and **zeros** are useful with standard models because a parameter or set of weights can be multiplied with the indicator feature. <br>
Consequently, <br> $0 *  value = 0$ <br> $1 * value = value$<br>
Functions may simply sum these indicator feature values to create statistics for indicator features, or to separate indicator features in the dataset.
